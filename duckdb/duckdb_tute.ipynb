```python
import duckdb

## Default storage is in-memory

## r is lazy
r  = duckdb.sql("SELECT 42")
print(r)

## storing in variable as table
s = duckdb.sql("SELECT 42 as i")
duckdb.sql("SELECT i * 2 AS k FROM s").show()

## Reading json
duckdb.read_json("data.json").show()
duckdb.sql("SELECT * FROM data.json")
```

```python
import pandas as pd

## Creating dataframes.
## This is readonly

df = pd.DataFrame({"a": [1]})
duckdb.sql("SELECT * FROM df")

```

```python
## Persisting to a file
import duckdb

## Insert data
con = duckdb.connect("data.db")
con.sql("CREATE TABLE  IF NOT EXISTS test (i INTEGER)")
con.sql("INSERT INTO test VALUES (42)")

con.table("test").show()
```

```python
## Read again the inserted data from before
con = duckdb.connect("data.db")
con.table("test").show()
```
